## 1. Question

## Answer 
*Can search console analytics history prioritize which content pages need updates first? We target content editors who make refresh decisions, helping them prioritize their limited capacity towards high-demand pages in decline.*

## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

## Answer
*I used the flyrank anonymized starter dataset (content_refresh_anonymized.csv), containing 30,000 paged and 32 distinct clients excluding

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

## Answer
*Defining the target as is_declining_label (1 if impressions fell >20% in the last 30 days, else 0).Then trained a Logistic Regression model on 17 numeric features and 8 one-hot encoded categorical features. Later validated the model using a Client-Holdout Split (reserving 20% of clients for testing) and evaluated using Precision@50.*
